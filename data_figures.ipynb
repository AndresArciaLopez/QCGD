# Heatmaps and Scatter Plots for Cost Function

```python
def plot_qcgd_heatmap(fig, ax1, heatmap_dic, title):   

    energies = heatmap_dic["energies"]
    M_values = heatmap_dic["M_values"]
    G_values = heatmap_dic["G_values"]
    M_min = heatmap_dic["M_min"]
    G_min = heatmap_dic["G_min"]
    N = heatmap_dic["L"]
    k = heatmap_dic["k"]
    
    heatmap = ax1.imshow(
        energies,
        origin="lower",
        aspect="auto",
        extent=[
            M_values[0],
            M_values[-1],
            G_values[0],
            G_values[-1],
        ],
        cmap="RdBu",)

    # Highlight the selected point
    ax1.scatter(
        M_min,
        G_min,
        s=100,
        color="silver",
        edgecolor="silver",
        linewidth=1.5,
        zorder=5,
        label=(rf"$\mu={M_min:.2f}, \gamma={G_min:.2f}, E_{k} = {np.min(energies):.2f}$"),
    )

    # Lines connecting the selected point to the axes
    ax1.vlines(
        M_min,
        ymin=G_values[0],
        ymax=G_min,
        colors="silver",
        linestyles="--",
        linewidth=1.5,
    )

    ax1.hlines(
        G_min,
        xmin=M_values[0],
        xmax=M_min,
        colors="silver",
        linestyles="--",
        linewidth=1.5,
    )

    # ax1.legend(loc="lower right",fontsize=20,)
    ax1.text(
        0.97,
        0.97,
        (
            rf"$\mu={M_min:.2f}$" "\n\n"
            rf"$\gamma={G_min:.2f}$" "\n\n"
            rf"$E_k={np.min(energies):.2f}$"
        ),
        transform=ax1.transAxes,
        ha="right",
        va="top",
        multialignment="left",  # aligns the lines inside the box
        fontsize=18,
        linespacing=0.5,
        bbox=dict(
            boxstyle="round,pad=0.3",
            facecolor="white",
            edgecolor="0.5",
            alpha=0.85,
        ),
    )
    fig.colorbar(heatmap, ax=ax1, label=rf"Energy $E_{k}(\mu, \gamma)$")
    ax1.set_xlabel(r" Step size $\mu$")
    ax1.set_ylabel(r" Conjugate parameter $\gamma$")
    ax1.set_title(rf"$L={N}$, step {k}, {title}")
    print(np.min(energies))
    # print(f"MIN ENERGY for L={N}, step {k}, {title}:", min(energies))
    # fig.savefig(f"energies{N}/Energy heatmap L={N} at step {k} for {title}.png", dpi=300, bbox_inches="tight")
    # plt.close(fig)

    # fig.savefig(
    #     f"energies{N}/Energy against mu L={N} at step {k}.png",
    #     dpi=300,
    #     bbox_inches="tight",
    # )
    # plt.close(fig)

def plot_qcgd_compare(ax2, line_dic, title):
    # ---------------------------------------------------------
    # Energy against mu for gamma=0 and gamma=optimal
    # ---------------------------------------------------------

    # Find the sampled gamma value closest to zero
    M_values = line_dic["M_values"]
    M_min = line_dic["M_min"]
    G_min = line_dic["G_min"]
    energies_gamma_zero = line_dic["energies_gamma_zero"]
    energies_gamma_optimal = line_dic["energies_gamma_optimal"]
    energy_min = line_dic["energy_min"]
    N = line_dic["L"]
    k = line_dic["k"]

    ax2.plot(
        M_values,
        energies_gamma_zero,
        marker="o",
        linestyle="-",
        label=rf"$\gamma={0}$",
    )

    ax2.plot(
        M_values,
        energies_gamma_optimal,
        marker="s",
        linestyle="-",
        label=rf"$\gamma_{{\mathrm{{opt}}}}={G_min:.2f}$",
    )

    # Highlight the globally selected point
    ax2.scatter(
        M_min,
        energy_min,
        s=50,
        color="black",
        edgecolor="black",
        linewidth=1.5,
        zorder=5,
        label=rf"$\mu={M_min:.2f}$, $\Delta E_{k}$ = {np.min(energies_gamma_zero) - energy_min:.2f}",
    )
    ax2.set_xlim([min(M_values), (2/3)* max(M_values)])
    ax2.set_ylim([min(energies_gamma_optimal) - 0.1, max(energies_gamma_optimal) + 0.5])
    # Connect the selected point to the plot axes
    x_axis_min, _ = ax2.get_xlim()
    y_axis_min, _ = ax2.get_ylim()

    ax2.vlines(
        M_min,
        ymin=y_axis_min,
        ymax=energy_min,
        colors="red",
        linestyles="--",
        alpha=0.7,
    )

    ax2.hlines(
        energy_min,
        xmin=x_axis_min,
        xmax=M_min,
        colors="red",
        linestyles="--",
        alpha=0.7,
    )

    ax2.set_xlabel(r"Step size $\mu$")
    ax2.set_ylabel(rf"Energy $E_{k}(\mu)$")
    # ax2.set_xlim(0, 0.4)
    # ax2.set_ylim(min(energies_gamma_optimal) - 0.1, max(energies_gamma_optimal) - 1.5)
    ax2.set_title(rf"$L={N}$, step {k}, {title}")
    ax2.grid(True, alpha=0.3)
    ax2.legend(loc="upper left", fontsize=20, ncol=1)

def plot_qite_steps(ax, line_dic, title):
    s_values = line_dic["s_values"]
    energies = line_dic["energies"]
    s_min = line_dic["s_min"]
    min_energy = line_dic["min_energy"]
    N = line_dic["L"]
    k = line_dic["k"]
    
    ax.plot(
        s_values,
        energies,
        marker="o",
        linestyle="-",
    )
    # Highlight the selected point
    ax.scatter(
        s_min,
        min_energy,
        s=50,
        color="black",
        edgecolor="black",
        zorder=5,
        label=rf"Selected: $\mu={s_min:.3f}$, Energy={min_energy:.3f}",
    )

    ax.set_xlim([min(s_values), max(s_values)])
    ax.set_ylim([min(energies) - 0.1, max(energies) + 0.1])
    # Get the plot limits so the lines stop at the axes
    x_min, x_max = ax.get_xlim()
    y_min, y_max = ax.get_ylim()

    ax.vlines(
        s_min,
        ymin=y_min,
        ymax=min_energy,
        colors="red",
        linestyles="--",
        alpha=0.7,
    )

    ax.hlines(
        min_energy,
        xmin=x_min,
        xmax=s_min,
        colors="red",
        linestyles="--",
        alpha=0.7,
    )

    ax.set_xlabel(r"$\mu$")
    ax.set_ylabel(rf"$\langle H \rangle_{{\psi_{k}}}$")
    ax.set_title(rf"L={N}, step {k}, {title}")
    ax.grid(True, alpha=0.3)
    ax.legend()
    # fig.savefig(
    #     f"energies{N}/Energies for L={N} at step {k} for {title} state.png",
    #     dpi=300,
    #     bbox_inches="tight",
    # )
    # plt.close(fig)
```

```python
run_param_rc(23)

L = 20
state = "HVA"


def load_landscape_data(x):
    base_folder = ["alpha1beta1", f"resultsL={L}"]

    return [
        load_data(
            f"EnergyL={L}step2{state}",
            folder=base_folder + [f"energies{L}range{x}x{x}"],
        ),
        load_data(
            f"EnergyL={L}step3{state}",
            folder=base_folder + [f"energies{L}range{x}x{x}"],
        ),
        load_data(
            f"EnergyL={L}step2{state}",
            folder=base_folder + [f"energies{L}mu{x}compare"],
        ),
        load_data(
            f"EnergyL={L}step3{state}",
            folder=base_folder + [f"energies{L}mu{x}compare"],
        ),
    ]


def add_panel_label(ax, label):
    ax.text(
        -0.14,
        1.04,
        f"({label})",
        transform=ax.transAxes,
        fontsize=24,
        fontweight="bold",
        ha="left",
        va="bottom",
        clip_on=False,
    )


restricted_data = load_landscape_data(0.5)
extended_data = load_landscape_data(1.5)

fig, axes = plt.subplots(
    2,
    4,
    figsize=(30, 11),
    width_ratios=[1, 1, 1, 1],
    constrained_layout=True,
)

# Restricted domain [0, 0.5]^2: panels (a)-(d)
plot_qcgd_heatmap(
    fig, axes[0, 0], restricted_data[0], title=state
)
plot_qcgd_compare(
    axes[0, 1], restricted_data[2], title=state
)
plot_qcgd_heatmap(
    fig, axes[1, 0], restricted_data[1], title=state
)
plot_qcgd_compare(
    axes[1, 1], restricted_data[3], title=state
)

# Extended domain [0, 1.5]^2: panels (e)-(h)
plot_qcgd_heatmap(
    fig, axes[0, 2], extended_data[0], title=state
)
plot_qcgd_compare(
    axes[0, 3], extended_data[2], title=state
)
plot_qcgd_heatmap(
    fig, axes[1, 2], extended_data[1], title=state
)
plot_qcgd_compare(
    axes[1, 3], extended_data[3], title=state
)

# Labels are not assigned with axes.flat because the two parameter
# domains are grouped by columns rather than in row-major order.
panel_labels = [
    ["a", "b", "e", "f"],
    ["c", "d", "g", "h"],
]

for row in range(2):
    for column in range(4):
        add_panel_label(axes[row, column], panel_labels[row][column])

plt.savefig(
    "plots/energy landscape combined.pdf",
    bbox_inches="tight",
    pad_inches=0.05,
)
plt.show()
```

# Excess Energy Ratio Table

```python
# Column headings
L_values = [10, 12, 14, 16, 18, 20]

# Replace these example values with your Python results.
# Each list must follow the order in L_values.
singlet_results = dict()

for i in [2,3]:
    ratios = []
    for l in L_values:
        energies_q = results_qcgd["singlet"][l]["qcgd_results"]["energies"]
        energies_db = results_qite["singlet"][l]["qite_results"]["energies"]
        energies_q_ratio = 100 * (1 - (energies_q[i] - E0_val[l])/(energies_q[0]-E0_val[l]))
        energies_db_ratio = 100 * (1- (energies_db[i]-E0_val[l])/(energies_db[0]-E0_val[l]))
        ratios.append(energies_q_ratio - energies_db_ratio)
    singlet_results[i] = ratios

HVA_results = dict()

for i in [2,3]:
    ratios = []
    for l in L_values:
        energies_q = results_qcgd["HVA"][l]["qcgd_results"]["energies"]
        energies_db = results_qite["HVA"][l]["qite_results"]["energies"]
        energies_q_ratio = 100 * (1 - (energies_q[i] - E0_val[l])/(energies_q[0]-E0_val[l]))
        energies_db_ratio = 100 * (1- (energies_db[i]-E0_val[l])/(energies_db[0]-E0_val[l]))
        ratios.append(energies_q_ratio - energies_db_ratio)
    HVA_results[i] = ratios

results = {
    "Singlet": {
        r"$\Delta r_2 \%$": singlet_results[2],
        r"$\Delta r_3 \%$": singlet_results[3],
    },
    "HVA": {
        r"$\Delta r_2 \%$": HVA_results[2],
        r"$\Delta r_3 \%$": HVA_results[3],
    },
}


def format_value(value):
    """Format table entries, including missing values."""
    if value is None:
        return "--"
    if isinstance(value, float):
        return f"{value:.1f}"
    return str(value)


def make_latex_table(results, L_values):
    n_numeric_columns = len(L_values)
    last_column = n_numeric_columns + 2

    column_spec = "|c|l|" + "c|" * n_numeric_columns
    header = " & $L$ & " + " & ".join(map(str, L_values)) + r" \\"

    rows = [
        rf"\begin{{tabular}}{{{column_spec}}}",
        r"\hline",
        header,
        r"\hline",
    ]

    for group_index, (group_name, quantities) in enumerate(results.items()):
        quantity_items = list(quantities.items())
        number_of_rows = len(quantity_items)

        for row_index, (quantity, values) in enumerate(quantity_items):
            if len(values) != n_numeric_columns:
                raise ValueError(
                    f"{group_name!r}, {quantity!r} contains {len(values)} "
                    f"values; expected {n_numeric_columns}."
                )

            formatted_values = " & ".join(format_value(x) for x in values)

            if row_index == 0:
                group_cell = (
                    rf"\multirow{{{number_of_rows}}}{{*}}{{{group_name}}}"
                )
            else:
                group_cell = ""

            rows.append(
                f"{group_cell} & {quantity} & {formatted_values} " + r"\\"
            )

            if row_index < number_of_rows - 1:
                # Horizontal line excluding the merged group cell.
                rows.append(rf"\cline{{2-{last_column}}}")

        rows.append(r"\hline")

    rows.append(r"\end{tabular}")

    return "\n".join(rows)


table = make_latex_table(results, L_values)

# Save just the table, so it can be inserted with \input{results_table.tex}.
Path("results_table.tex").write_text(table, encoding="utf-8")

print(table)
```

```python
from matplotlib.lines import Line2D

final_fidelity = {"qcgd": {"HVA": {}, "singlet": {}}, "dbqite": {"HVA": {}, "singlet": {}}}
def generate_plots(ax1, L, key):
    
    lables_energies={"singlet": r"$E_k$ Singlet", "HVA": r"$E_k$ HVA"}
    lables_fidelities={"singlet": r"$F_k$ Singlet", "HVA": r"$F_k$ HVA"}
    alphas = [1.0, 0.6]
    markers = ['d','o']
    indices = ["HVA","singlet"] 
    steps = [0,1,2,3]
    colors = {"qcgd":'#20306f',"dbqite":'#6929C4'}

    if key == "qcgd":
        results = results_qcgd
        dict_key = "qcgd_results"
    elif key == "dbqite":
        results = results_qite  
        dict_key = "qite_results"
    else:
        raise ValueError("Invalid key. Must be 'qcgd' or 'dbqite'.")
    
    ###############################################################################
    # SUBPLOT 1: Energy
    ###############################################################################

    # 1) Plot energies on the left y-axis
    if key == "qcgd":
            facecolor = 'none'
    else:
        facecolor = colors[key]
    for i,ind in enumerate(indices):
        line1, = ax1.plot(steps, results[ind][L][dict_key]['energies'], label=lables_energies[ind], color=colors[key], mfc=facecolor, marker=markers[i], linestyle='solid', linewidth=2, zorder=3, alpha=alphas[i])

    ax1.axhline(y=E0_val[L], color="red", linestyle=':', alpha=1.0)
    ax1.axhline(y=E1_val[L], color="red", linestyle=':', alpha=0.6)

    ax1.text(
        0.2,           
        E0_val[L]+0.1,       
        r"$\lambda_0$",                          
        va="center",
        ha="left",
        color="red",
    )   

    ax1.text(
        0.2,           
        E1_val[L]+0.1,             
        r"$\lambda_1$",                           
        va="center",
        ha="left",
        color="red",
        alpha = 0.6
    )

    ax1.set_xlabel(r"Steps $k$")
    ax1.set_ylabel(r"Energy $E_k$")
    ax1.set_xticks(ticks = range(0, 6, 1))
    ax1.set_title(f"QCGD and DB-QITE for $L={L}$")

    ###############################################################################
    # SUBPLOT 2: Fidelity
    ###############################################################################

    # 2) Create a twin axis for fluctuations
    axF = ax1.twinx()

    # 1) Plot fidelities on the left y-axis
    if key == "qcgd":
        facecolor = 'none'
    else:
        facecolor = colors[key]
    for i,ind in enumerate(indices):
        line1, = axF.plot(steps, results[ind][L]['fidelities'], label=lables_fidelities[ind], color=colors[key], mfc=facecolor, marker=markers[i], linestyle='solid', linewidth=2, zorder=3, alpha=alphas[i])

    axF.axhline(y=1, color="orange", linestyle=':', alpha=1.0)
    axF.axhline(y=0.975, color="orange", linestyle=':', alpha=0.8)
    axF.axhline(y=0.95, color="orange", linestyle=':', alpha=0.6)

    axF.set_ylabel(r"Fidelity $F_k$")

    ###############################################################################

    # 3) Combine legends from both axes
    # lines_1, labels_1 = ax1.get_legend_handles_labels()
    # lines_2, labels_2 = axF.get_legend_handles_labels()
    # ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc="center left", ncol=2, fontsize=12)

    final_fidelity[key]["HVA"][L] = np.round(results["HVA"][L]['fidelities'][-1], 3)
    final_fidelity[key]["singlet"][L] = np.round(results["singlet"][L]['fidelities'][-1], 3)
    if key == "qcgd":
        from matplotlib.lines import Line2D
        legend_handles = [
            Line2D(
                [0], [0],
                marker='d',
                color='black',
                markerfacecolor='none',
                markersize=8,
                label='HVA',
            ),
            Line2D(
                [0], [0],
                marker='o',
                color='black',
                markerfacecolor='none',
                markersize=8,
                label='singlet',
            ),
        ]
        ax1.legend(handles=legend_handles, loc='center left', fontsize=12, title='State', title_fontsize='13')
def generate_fidelity_latex_table(
    final_fidelity,
    L_values=None,
    decimals=3,
    caption="Final ground-state fidelities after three iterations.",
    label="tab:final_fidelities",
):
    if L_values is None:
        L_values = sorted({
            L
            for method_data in final_fidelity.values()
            for initialisation_data in method_data.values()
            for L in initialisation_data
        })

    def format_value(method, initialisation, L):
        value = final_fidelity[method][initialisation].get(L)

        if value is None:
            return "--"

        return f"{float(value):.{decimals}f}"

    rows = [
        r"\begin{table}[t]",
        r"\centering",
        rf"\caption{{{caption}}}",
        rf"\label{{{label}}}",
        r"\begin{tabular}{c cc cc}",
        r"\toprule",
        (
            r"$L$"
            r" & \multicolumn{2}{c}{DB-QITE}"
            r" & \multicolumn{2}{c}{QCGD} \\"
        ),
        r"\cmidrule(lr){2-3}\cmidrule(lr){4-5}",
        (
            r" & HVA & Singlet"
            r" & HVA & Singlet \\"
        ),
        r"\midrule",
    ]

    for L in L_values:
        row = (
            rf"{L}"
            rf" & {format_value('dbqite', 'HVA', L)}"
            rf" & {format_value('dbqite', 'singlet', L)}"
            rf" & {format_value('qcgd', 'HVA', L)}"
            rf" & {format_value('qcgd', 'singlet', L)} \\"
        )
        rows.append(row)

    rows.extend([
        r"\bottomrule",
        r"\end{tabular}",
        r"\end{table}",
    ])

    return "\n".join(rows)
```

```python
import matplotlib


def generate_circuit_plots(axC, L, key, init_type='HVA'):

    ###############################################################################
    # SUBPLOT 3: Circuit data 
    ###############################################################################
    if key == "qcgd":
        results = results_qcgd
        dict_key = "qcgd_results"
        cost_colors = {"U3 count": "#20306F","CZ count": "#4776B4","Depth": "#258B9B"}

    elif key == "dbqite":
        results = results_qite  
        dict_key = "qite_results"
        cost_colors = {"U3 count": "#6929C4","CZ count": "#9A5BD5","Depth": "#C58BE2"}
    else:
        raise ValueError("Invalid type. Must be 'qcgd' or 'qite'.")

    iterations = len(results["singlet"][L][dict_key]["energies"])
    steps = range(0, iterations)

    # Singlet
    u_count_S = np.array([results["singlet"][L][dict_key]["circuit_data"][0][k]['u3'] for k in steps])
    cz_count_S = np.array([results["singlet"][L][dict_key]["circuit_data"][0][k]['cz'] for k in steps])
    depth_S = np.array([results["singlet"][L][dict_key]["circuit_data"][2][k] for k in steps])

    # HVA
    u_count_HVA = np.array([results["HVA"][L][dict_key]["circuit_data"][0][k]['u3'] for k in steps])
    cz_count_HVA = np.array([results["HVA"][L][dict_key]["circuit_data"][0][k]['cz'] for k in steps])
    depth_HVA = np.array([results["HVA"][L][dict_key]["circuit_data"][2][k] for k in steps])
    scaling = 1000

    if init_type=='Singlet':
        axC.plot(steps, u_count_S/scaling, marker='d', color=cost_colors["U3 count"], alpha=1.0, label='U3 count')
        axC.plot(steps, cz_count_S/scaling, marker='o', color=cost_colors["CZ count"], alpha=1.0, label='CZ count')
        axC.plot(steps, depth_S/scaling, marker='s', color=cost_colors["Depth"], alpha=1.0, label='Depth')

        axC.plot(steps, u_count_HVA/scaling, color=cost_colors["U3 count"], marker='d', alpha=0.6, linestyle='dashed') 
        axC.plot(steps, cz_count_HVA/scaling, color=cost_colors["CZ count"], marker='o', alpha=0.6, linestyle='dashed') 
        axC.plot(steps, depth_HVA/scaling, color=cost_colors["Depth"], marker='s', alpha=0.6, linestyle='dashed')    
    else:
        axC.plot(steps, u_count_HVA/scaling, marker='d', color=cost_colors["U3 count"], alpha=1.0, label='U3 count')
        axC.plot(steps, cz_count_HVA/scaling, marker='o', color=cost_colors["CZ count"], alpha=1.0, label='CZ count')
        axC.plot(steps, depth_HVA/scaling, marker='s', color=cost_colors["Depth"], alpha=1.0, label='Depth')

        axC.plot(steps, u_count_S/scaling, color=cost_colors["U3 count"], marker='d', alpha=0.6, linestyle='dashed') 
        axC.plot(steps, cz_count_S/scaling, color=cost_colors["CZ count"], marker='o', alpha=0.6, linestyle='dashed') 
        axC.plot(steps, depth_S/scaling, color=cost_colors["Depth"], marker='s', alpha=0.6, linestyle='dashed')        

    axC.set_xlabel(r"Steps $k$")
    axC.set_ylabel("Counts in thousands")
    axC.set_xticks(ticks = steps)

    axC.legend(loc="upper left", ncols=3, fontsize=15)
    if key == "dbqite":
        title = "DB-QITE"
    else:
        title = "QCGD"

    axC.set_title(f"{title} for $L={L}$")

    ###############################################################################
    # Inset table: circuit data GC
    ###############################################################################
    fidelities = np.round(results["singlet"][L]["fidelities"], 3) if init_type=='Singlet' else np.round(results["HVA"][L]["fidelities"], 3)

    # Data for the table
    rows = [str(k) for k in steps]
    columns = (r'$k$','U3', 'CZ', 'Depth', r'$|\langle \psi_k | \lambda_0 \rangle |^2$')
    if init_type=='Singlet':
        cell_text = [[rows[i], u_count_S[i], cz_count_S[i], depth_S[i], fidelities[i]] for i in range(len(rows))]
    else:
        cell_text = [[rows[i], u_count_HVA[i], cz_count_HVA[i], depth_HVA[i], fidelities[i]] for i in range(len(rows))]

    # Adjust column widths
    col_widths = [0.05, 0.1, 0.1, 0.1, 0.1]  # First column is wider

    # Add a table at the bottom of the axes
    tableC = axC.table(cellText=cell_text,
                        colLabels=columns,
                        loc='center left',
                        colWidths=col_widths)


    tableC.auto_set_font_size(False)
    tableC.set_fontsize(12)
    tableC.scale(1.5, 1.5)  # Adjust the scaling as needed
    # Move the entire table upward by 5 pixels
    tableC.set_transform(
    tableC.get_transform()
    + matplotlib.transforms.ScaledTranslation(
        -0 / 72, 20 / 72, fig.dpi_scale_trans
    )
)
```

# Comparison DB-QITE vs QCGD


```python
markers = ['d', '^', 'v', '<', '>', 'o']
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']
alphas = [0.25,0.4,0.55,0.7,0.85,1]

def generate_fidelity_cz(ax, init_type, key, show_legend=True):
    if key == "qcgd":
        results = results_qcgd
        dict_key = "qcgd_results"
    elif key == "dbqite":
        results = results_qite  
        dict_key = "qite_results"
    else:
        raise ValueError("Invalid type. Must be 'qcgd' or 'dbqite'.")

    if init_type=="singlet":
        ax.set_title(rf"{key.upper()} for $|$Singlet$\rangle$")
    if init_type=="HVA":
        ax.set_title(rf"{key.upper()} for $|$HVA$\rangle$")

    steps = list(range(0,4))

    cz_data_qite = dict()
    for i,L in enumerate(list(range(10,22,2))):
        cz = np.array([results[init_type][L][dict_key]["circuit_data"][0][k]['cz'] for k in steps])
        cz_data_qite[L] = cz

    for i,L in enumerate(list(range(10,22,2))):
        fidelities = results[init_type][L]["fidelities"]
        cz_counts = cz_data_qite[L]
        ax.plot(cz_counts, fidelities, marker=markers[i], label=r'$L=$'+str(L), color= colors[0], alpha = alphas[i])

    ax.axhline(y=1, color="orange", linestyle=':', alpha=1.0)
    ax.axhline(y=0.975, color="orange", linestyle=':', alpha=0.8)
    ax.axhline(y=0.95, color="orange", linestyle=':', alpha=0.6)

    if show_legend:
        ax.legend(loc="best", ncols=2, fontsize=15)
    ax.set_xscale('log')
    ax.set_xlabel(r"Number of CZ gates")
    ax.set_ylabel(r"Fidelity $F_k$")
```

```python
run_param_rc(23)

fig, ((axS_QITE, axS_QCGD), (axV_QITE, axV_QCGD)) = plt.subplots(2, 2, figsize=(12, 10), gridspec_kw={"width_ratios": [1,1]})

generate_fidelity_cz(axS_QITE, "singlet", "dbqite")
generate_fidelity_cz(axV_QITE, "HVA", "dbqite" ,show_legend=False)
generate_fidelity_cz(axS_QCGD, "singlet", "qcgd")
generate_fidelity_cz(axV_QCGD, "HVA", "qcgd", show_legend=False)

plt.tight_layout()
#plt.savefig("plots/QITE_QPE_Heisenberg.pdf")
plt.show()
```
